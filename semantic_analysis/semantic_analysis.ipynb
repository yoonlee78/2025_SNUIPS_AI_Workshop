```python
from konlpy.tag import Mecab
mecab = Mecab()
```

Mecab에 내재된 함수를 이용해 형태소 분석을 시행합니다.

본 워크샵에서는 여러 형태소 중, 유의미한 내용어만을 대상으로 워드 임베딩을 진행할 것이므로 mecab.nouns()를 이용해 명사만 추출한 데이터를 따로 만듭니다.

```python
# 형태소 분석 실행
text_all_df["morphs"] = text_all_df["text_all"].astype(str).apply(lambda x: mecab.morphs(x))  # 형태소 추출
text_all_df["nouns"] = text_all_df["text_all"].astype(str).apply(lambda x: mecab.nouns(x)) # 명사 추출
```

```python
text_all_df.head()
```

- 추출된 명사 중에서도 정보값이 적은 전치사, 대명사, 감탄사 등을 불용어(stop words)로 지정하여 제외합니다.

```python
# 초성 제거 함수
import re
def remove_chosung(text):
  return re.sub(r"[ㄱ-ㅎ]", "", text)

# 불용어 리스트 (조사, 감탄사, 의존 명사 등)
stop_words = set(["이", "그", "저", "이것", "저것", "그것", "이게", "그게", "저게", "그거", "저거", "이거",
    "그건", "전", "만", "정도",
    "나", "난", "내", "네", "내가", "너", "넌", "네가", "당신", "우리", "너희", "그들", "저희",
    "에게", "에서", "에게서", "로부터", "로써", "로서", "같이", "처럼", "보다", "까지", "만큼",
    "의", "에", "에서", "로", "에게", "도", "는", "은", "를", "을", "가", "이", "와", "과",
    "한테", "밖에", "마다", "조차", "라든가", "더러", "던가", "인데", "라도", "며", "하고", "것", "거", "게", "지", "데", "등", "중", "바", "줄", "경우", "동안",
    "이야", "따름", "뿐", "대로", "만큼", "따위", "때문", "한", "걸", "건", "것", "누구", "듯", "수", "아", "야", "어", "오", "우와", "와", "아하", "어머", "어머나", "아이구", "아이코",
    "때", "일", "날", "번", "헉", "헐", "허", "흠", "휴", "음", "에이", "에구", "에휴", "참", "뭐", "뭐야", "뭐라구"])

# 불용어 제거 적용
text_all_df["filtered_nouns"] = text_all_df["nouns"].apply(lambda words: [remove_chosung(word) for word in words if word not in stop_words])
```

```python
text_all_df[["filtered_nouns"]].head()
```

- 이제 필터링된 모든 명사를 토큰으로 바꿔줍니다.

```python
tokenized_nouns = text_all_df["filtered_nouns"].tolist()
```

```python
print(tokenized_nouns[0])
```

```python
token_counts = [len(sublist) for sublist in tokenized_nouns]
total_tokens = sum(token_counts)

# 결과 출력
print("각 리스트별 토큰 수:", token_counts)
print("전체 토큰 수:", total_tokens)
```

## Build Word2Vec Model

텍스트 데이터 전처리가 완료되었으므로, Word2Vec 모델을 생성합니다. Word2Vec 모델 학습을 위해 권장되는 토큰 수는 수십만 개에서 수백만 개이나, 현재는 공개된 데이터셋으로 모델을 만들고 있는 점을 감안하여 17만 개 토큰으로 모델을 만들어보고자 합니다.

```python
from gensim.models import Word2Vec

# Word2Vec 모델 학습 (Skip-gram 사용)
model = Word2Vec(
    sentences=tokenized_nouns,  # 입력 데이터
    vector_size=100,  # 벡터 차원 수
    window=4,  # 컨텍스트 윈도우 크기
    min_count=3,  # 최소 등장 횟수 (이보다 적게 등장한 단어는 무시)
    workers=4,  # 병렬 처리에 사용할 CPU 스레드 개수
    epochs=20,
    sg=1,  # Skip-gram 사용 (1: Skip-gram, 0: CBOW)
    negative=8
)
```

```python
len(model.wv)
```

- 워드 임베딩 결과를 확인해봅시다! most_similar() 안에 학습된 단어 중 하나를 넣으면, 그와 가장 유사한 단어 목록이 제공됩니다.

```python
model.wv.most_similar('노래')
```

## Clustering
워드 임베딩 결과를 수동으로 확인하는 데에는 한계가 있으니, K-Means clustering을 이용해 생성된 워드 임베딩을 클러스터화해서 시각화하고자 합니다.

```python
import numpy as np
from sklearn.cluster import KMeans

words = list(model.wv.index_to_key)
word_vectors = np.array([model.wv[word] for word in words])

kmeans = KMeans(n_clusters=6) # 클러스트 개수를 지정하는 행으로, 이 개수는 연구자 자율에 따릅니다.
kmeans.fit(word_vectors)

word_clusters = {word: kmeans.labels_[i] for i, word in enumerate(words)}

clustered_words_top20 = {}
for cluster in range(6):
    print(f"\n📌 클러스터 {cluster+1} :")
    clustered_words = [word for word, label in word_clusters.items() if label == cluster]
    clustered_words_top20[cluster] = clustered_words[:20]
    print(clustered_words[:20])  # 상위 20개 단어만 출력
```

- Google Colab의 matplot은 기본적으로 한국어를 지원해주지 않아 koreanize_matplotlib을 설치합니다.

```python
!pip install koreanize-matplotlib
import koreanize_matplotlib
import matplotlib.pyplot as plt
```

```python
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE

# t-SNE로 2차원 축소
tsne = TSNE(n_components=2, random_state=42, perplexity=30)
word_vectors_2d = tsne.fit_transform(word_vectors)

# 클러스터별 색상 설정
colors = plt.cm.get_cmap("tab10", 6)  # 6개의 클러스터 색상 지정

plt.figure(figsize=(12, 6))
for cluster_id in range(6):
    # 클러스터별 상위 20개 단어 가져오기
    cluster_words = clustered_words_top20[cluster_id]
    valid_cluster_words = [word for word in cluster_words if word in words]

    # 단어의 인덱스와 좌표 추출
    indices = [words.index(word) for word in valid_cluster_words]
    cluster_points = word_vectors_2d[indices]

    # 클러스터 시각화
    if len(cluster_points) > 0:  # 데이터가 있을 경우에만 시각화
        plt.scatter(cluster_points[:, 0], cluster_points[:, 1], color=colors(cluster_id), label=f'클러스터 {cluster_id+1}')

        # 텍스트 추가
        for word, (x, y) in zip(valid_cluster_words, cluster_points):
            plt.text(x, y, word, fontsize=9, alpha=0.7)

plt.legend()
plt.title("워드 클러스터링 시각화 (상위 20개 단어)")
plt.xlabel("t-SNE 차원 1")
plt.ylabel("t-SNE 차원 2")
plt.show()
```

# LDA Topic Modeling
이제 같은 데이터셋에 LDA Topic Modeling을 시행하고자 합니다. 이를 위해서는 gensim library를 import해야 합니다.

```python
import gensim
```

이제 LDA에 넣기 위한 코퍼스를 만들어보겠습니다. 이때 코퍼스에 들어갈 단어는 앞서 생성한, tokenized_nouns입니다.

```python
dictionary = gensim.corpora.Dictionary(tokenized_nouns)
corpus = [dictionary.doc2bow(text) for text in tokenized_nouns]
```

다음으로 LDA 모델링을 직접 시행하겠습니다.
K-Means clustering과 비슷하게 몇 개의 주제를 추출할지 먼저 지정해주어야 합니다.

```python
from gensim.models import LdaModel

# LDA 모델 학습
num_topics = 6  # 추출할 주제 개수
lda_model = LdaModel(corpus=corpus, num_topics=num_topics,
                     id2word=dictionary, passes=10, random_state=42)

# 결과 출력
topics = lda_model.print_topics(num_words=5)  # 각 주제에서 상위 5개 단어 출력
for idx, topic in topics:
    print(f"주제 {idx}: {topic}")
```
